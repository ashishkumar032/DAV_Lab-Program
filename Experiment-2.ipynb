```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Asha", "Ravi", "Mina", "Ravi", "Leo", "Tara", "Asha", "Nikhil"],
    "Age": [21, 23, None, 23, 25, 22, 21, None],
    "City": ["Delhi", "Mumbai", "Pune", "Mumbai", "Delhi", "Chennai", "Delhi", "Pune"],
    "Department": ["Sales", "IT", "HR", "IT", "Sales", "HR", "Sales", "IT"],
    "Salary": [35000, 48000, 42000, 48000, None, 39000, 35000, 51000],
    "Rating": [4.2, 3.8, 4.7, 3.8, 4.1, None, 4.2, 4.5]
})

print("First five rows:")
display(df.head())

print("Dataset size:", df.shape)
print("Column names:", df.columns.tolist())

print("Data types:")
print(df.dtypes)

print("Descriptive statistics:")
display(df.describe(include="all"))

print("Missing values in each column:")
print(df.isnull().sum())

print("Total missing values:", df.isnull().sum().sum())

print("Number of duplicate rows:", df.duplicated().sum())

clean_df = df.drop_duplicates()

print("Dataset after removing duplicates:")
display(clean_df)

print("Missing values after removing duplicates:")
print(clean_df.isnull().sum())
```
