```python
import numpy as np
import pandas as pd

numbers = np.array([5, 10, 15, 20, 25])
print("Array:", numbers)
print("Array + 5:", numbers + 5)
print("Array × 2:", numbers * 2)
print("First three values:", numbers[:3])
print("Values above 12:", numbers[numbers > 12])

data = {
    "Name": ["Asha", "Ravi", "Mina", "Arjun"],
    "Age": [21, 23, 20, 22],
    "Score": [88, 76, 95, 81]
}

df = pd.DataFrame(data)
print("\nDataFrame:")
print(df)

print("\nScore column:")
print(df["Score"])

print("\nFirst two rows:")
print(df.iloc[:2])

print("\nNames and scores for the first three rows:")
print(df.loc[:2, ["Name", "Score"]])

print("\nStudents with scores above 80:")
print(df[df["Score"] > 80])
```
