```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    "StudyHours": [2, 4, 5, 3, 7, 8, 6, 9, 1, 5],
    "Score": [55, 65, 72, 60, 82, 90, 78, 95, 48, 70],
    "Attendance": [60, 75, 80, 70, 90, 95, 85, 98, 55, 78],
    "SleepHours": [6, 7, 8, 6, 7, 8, 7, 8, 5, 7]
})

display(df.head())

sns.histplot(data=df, x="Score", bins=5, kde=True)
plt.title("Distribution of Scores")
plt.show()

sns.boxplot(data=df, x="Score")
plt.title("Score Box Plot")
plt.show()

sns.scatterplot(data=df, x="StudyHours", y="Score", hue="Attendance")
plt.title("Study Hours vs Score")
plt.show()

sns.pairplot(df)
plt.show()

sns.heatmap(df.corr(), annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.show()
```
